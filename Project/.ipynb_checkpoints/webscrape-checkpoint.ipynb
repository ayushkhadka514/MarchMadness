```python
import requests
from bs4 import BeautifulSoup
import re
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
import os
```

```python
url = "https://barttorvik.com/trank.php?year=2008#"

headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)

# Check if the request was successful
if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")

    # Find table rows with March Madness teams (class="seedrow")
    rows = soup.find_all("tr", class_="seedrow")

    # Extract Data
    data = []
    for row in rows:
        cols = [td.text.strip() for td in row.find_all("td")]

        # Ensure we extract relevant data
        if len(cols) > 6:  # Ensure it has enough columns
            team_name = row.get("id", "Unknown")  
            seed = re.search(r"(\d+) seed", cols[1])  # Extract seed number

            data.append([
                team_name,  # Team
                cols[2],  # Conf
                cols[3],  # Games Played
                cols[4],  # Wins
                cols[5],  # ADJOE
                cols[6],  # ADJDE
                cols[7],  # BARTHAG
                cols[8],  # EFG_O
                cols[9],  # EFG_D
                cols[10], # TOR
                int(seed.group(1)) if seed else None  # Extracted Seed
            ])

    df = pd.DataFrame(data, columns=["Team", "Conf", "G", "W", "ADJOE", "ADJDE", "BARTHAG", "EFG_O", "EFG_D", "TOR", "Seed"])

else:
    print("Failed to retrieve the webpage.")

df["W"] = df["W"].apply(lambda x: re.search(r"^\d+", x).group() if re.search(r"^\d+", x) else x)
df = df.dropna(subset=["Seed"])
# Save to CSV
df.to_csv("cbb08.csv", index=False)

df.head(100)
```

```python
path = 'data/'
csv_files = [file for file in os.listdir(path) if file.endswith(".csv")]

dataframes = []

for file in csv_files:
    file_path = os.path.join(path, file)
    
    df = pd.read_csv(file_path)
    
    year = '20' + file[3:5]
    
    df["Team"] = df["Team"].str.lstrip("_")  
    df["Team"] = year + "_" + df["Team"]  

    dataframes.append(df)

combined_df = pd.concat(dataframes, ignore_index=True)

combined_df.to_csv("pastCBB.csv", index=False)

df = pd.read_csv('pastCBB.csv')
df.head(10)
```

```python
existing_df = pd.read_csv('cbb25.csv')

# Web scraping
url = "https://barttorvik.com/trank.php?year=2025#"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    rows = soup.find_all("tr", class_="seedrow")

    new_data = []

    for row in rows:
        cols = [td.text.strip() for td in row.find_all("td")]
        if len(cols) > 23:
            team_name = row.get("id", "Unknown")

            # Get only the new columns
            new_data.append([
                team_name,
                cols[11],  # TORD
                cols[12],  # ORB
                cols[13],  # DRB
                cols[14],  # FTR
                cols[15],  # FTRD
                cols[16],  # 2P%
                cols[17],  # 2P%D
                cols[18],  # 3P%
                cols[19],  # 3P%D
                cols[20],  # 3PR
                cols[21],  # 3PRD
                cols[22],  # ADJ_T
                cols[23]   # WAB
            ])

    extra_cols = ["Team", "TORD", "ORB", "DRB", "FTR", "FTRD", "2P%", "2P%D",
                  "3P%", "3P%D", "3PR", "3PRD", "ADJ_T", "WAB"]
    new_df = pd.DataFrame(new_data, columns=extra_cols)

    # Merge on Team name
    full_df = pd.merge(existing_df, new_df, on="Team", how="left")

    # Save combined file
    full_df.to_csv("cbb25_full_combined.csv", index=False)
    print(full_df.head())

else:
    print("Failed to fetch the webpage.")
```
